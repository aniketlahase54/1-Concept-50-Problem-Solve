```python
# Check whether a number is prime.

num = 7
is_prime = True

if num < 2:
    is_prime = False
else:
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

if is_prime:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
# Student Marks: Given:
# marks = [85, 72, 90, 65, 78]

# Use a loop to calculate:

# Total
# Average
# Highest marks
# Lowest marks
# Student Marks

marks = [85, 72, 90, 65, 78]

total = 0
count = 0
high = marks[0]
low = marks[0]

for i in marks:
    total += i
    count += 1

    if i > high:
        high = i

    if i < low:
        low = i

average = total / count

print("Total:", total)
print("Average:", average)
print("Highest Marks:", high)
print("Lowest Marks:", low)
```

```python
# String Analyzer: Given:
# text = "Python Programming"

# Use a loop to count:

# Total characters
# Vowels
# Consonants
# Spaces


text = "Python Programming"

total_characters = 0
vowels = 0
consonants = 0
spaces = 0

for i in text:
    total_characters += 1

    if i in "aeiouAEIOU":
        vowels += 1

    elif i == " ":
        spaces += 1

    else:
        consonants += 1

print("Total Characters:", total_characters)
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Spaces:", spaces) 
```

```python
# 🚀 Mini Project — Data Science Practice

# Given:

# scores = [85, 72, 91, 64, 78, 95, 55, 88]

# Use loops to find:

# Highest score
# Lowest score
# Total score
# Average score
# Number of students who scored 80+
# Number of students who passed (40+)
# Number of students who failed (<40)

scores = [85, 72, 91, 64, 78, 95, 55, 88]

total = 0
count = 0

highest = scores[0]
lowest = scores[0]

students_80_plus = 0
passed = 0
failed = 0

for score in scores:

    # Total
    total += score
    count += 1

    # Highest
    if score > highest:
        highest = score

    # Lowest
    if score < lowest:
        lowest = score

    # 80+
    if score >= 80:
        students_80_plus += 1

    # Pass / Fail
    if score >= 40:
        passed += 1
    else:
        failed += 1

# Average
average = total / count

print("Highest Score:", highest)
print("Lowest Score:", lowest)
print("Total Score:", total)
print("Average Score:", average)
print("Students Scored 80+:", students_80_plus)
print("Students Passed:", passed)
print("Students Failed:", failed)
```
