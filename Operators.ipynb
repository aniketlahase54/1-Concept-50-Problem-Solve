```python
# 🔵 Identity Operators — 42–45
# Check whether two variables refer to the same object using is.
a = [1, 2, 3]
b = a

print(a is b)
```

```python
# Create two lists and observe the difference between == and is.
list1 = [10, 20, 30]
list2 = [10, 20, 30]

print(list1 == list2)  # Checks values
print(list1 is list2)  # Checks object identity
```

```python
# Assign one list to another variable and check the result using is.
list1 = [10, 20, 30]
list2 = list1

print(list1 is list2)
```
