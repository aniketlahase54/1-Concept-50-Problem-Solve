```python
# Create a BankAccount class with deposit() and withdraw() instance methods.
class BankAccount: 
    def __init__(self,balc): 
        self.balc = balc

    def deposit(self,amount):
        self.balc+=amount 
        print("Deposit successeful")

    def withdraw(self,amount):
        self.balc-=amount
        print("Withdraw successful")
        
b1 = BankAccount(15000)
b1.deposit(5000)
print(b1.balc)
b1.withdraw(10000)
print(b1.balc)
```

```python
# Create a Calculator class with instance methods for addition, subtraction, multiplication, and division.

class Calculator:
    def __init__(self,a,b): 
        self.a = a
        self.b = b 

    def add(self):
        return self.a + self.b

    def sub(self):
        return self.a - self.b
        
    def mul(self):
        return self.a * self.b
    
    def div(self):
        return self.a / self.b

c1 = Calculator(10,2)
print("Addition:",c1.add())
print("Subtraction:",c1.sub())
print("Multiplication:",c1.mul())
print("Division:",c1.div())
```

```python
# Create a MathHelper class with static methods to check whether a number is even, odd

class MathHelper: 

    @staticmethod
    def check(num): 
        if num%2==0: 
            print("EVEN")

        else: 
            print("ODD")
m1 = MathHelper()
m1.check(76)
```

```python
# Create a Student class containing:
# class variable
# instance variables
# parameterized constructor
# instance method
# class method
# static method

# Then create multiple objects and demonstrate how each type works.

class Student:

    # Class Variable
    school = "ABC School"

    # Parameterized Constructor
    def __init__(self, name, marks):
        # Instance Variables
        self.name = name
        self.marks = marks

    # Instance Method
    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)
        print("School:", self.school)

    # Class Method
    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school

    # Static Method
    @staticmethod
    def check_result(marks):
        if marks >= 40:
            print("PASS")
        else:
            print("FAIL")


# Multiple Objects
s1 = Student("Aniket", 90)
s2 = Student("Om", 75)
s3 = Student("Kartik", 35)


# Instance Method
s1.display()
print("----------------")

s2.display()
print("----------------")

s3.display()
print("----------------")


# Class Method
Student.change_school("XYZ School")

print(s1.school)
print(s2.school)
print(s3.school)
print("----------------")


# Static Method
Student.check_result(s1.marks)
Student.check_result(s3.marks)
```
