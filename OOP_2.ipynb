```python
# Create a Book class and use __str__() to display its title and author.
class book : 
    def __init__(self,title,author):
        self.title = title
        self.author = author

    def __str__(self): 
        return f"Title : {self.title} \nAuthor : {self.author}"

b1 = book("Python Crash Course","Eric Matthes")
print(b1)
```

```python
# Create a Product class and implement __repr__().
class product: 
    def __init__(self,name,price):
        self.name = name 
        self.price = price


    def __repr__(self): 
        return f'product("{self.name}", "{self.price}")'

p1 = product("pen",10)
print(p1)
```

```python
# Create a Student class and use __len__() to return the number of subjects.
class Student:
    def __init__(self, name, subjects):
        self.name = name
        self.subjects = subjects

    def __len__(self):
        return len(self.subjects)


s1 = Student("Aniket", ["Python", "SQL", "Machine Learning", "Statistics"])

print(s1.name)
print(len(s1))
```

```python
# Create a Number class and implement __add__() for adding two objects.
class Number: 
    def __init__(self,num):
        self.num = num
        

    def __add__(self,other):
        c = self.num + other.num
        return Number(c)

    def __str__(self):
        return str(self.num)
        
a = Number(5)
b = Number(3)
print(a + b)
```

```python
# Create a Point class and implement __eq__() to compare two objects.
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y


p1 = Point(10, 20)
p2 = Point(10, 20)
p3 = Point(5, 15)

print(p1 == p2)
print(p1 == p3)
```

```python
# Create a Student class and implement __lt__() to compare marks.
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __lt__(self, other):
        return self.marks < other.marks


s1 = Student("Aniket", 85)
s2 = Student("Rahul", 75)

print(s1 < s2)
print(s2 < s1)
```

```python
# Create a Counter class and implement __add__() between two counter objects.
class Counter:
    def __init__(self, count):
        self.count = count

    def __add__(self, other):
        return Counter(self.count + other.count)

    def __str__(self):
        return str(self.count)


c1 = Counter(10)
c2 = Counter(20)

c3 = c1 + c2

print(c3)
```

```python
# Create a private variable __salary and access it using a public getter method.
class Employee:
    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary


e1 = Employee(45000)
print("Salary:", e1.get_salary())
```

```python
# Create a BankAccount class with private __balance and create deposit() and get_balance() methods.
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance


account = BankAccount(5000)

account.deposit(2000)

print("Balance:", account.get_balance())
```

```python
# Final Challenge: Create a Student class containing:
# Public variable name
# Protected variable _marks
# Private variable __password
# __init__()
# __str__()
# Public method
# Protected method
# Private method
# Attribute creation outside the class
# Object reference
# Function receiving the object

# Then create multiple objects and demonstrate how each concept works.

class Student:

    def __init__(self, name, marks, password):
        self.name = name              # Public variable
        self._marks = marks           # Protected variable
        self.__password = password    # Private variable

    def __str__(self):
        return f"Name: {self.name}, Marks: {self._marks}"

    # Public method
    def display(self):
        print("Name:", self.name)
        print("Marks:", self._marks)

    # Protected method
    def _show_marks(self):
        print("Protected Method - Marks:", self._marks)

    # Private method
    def __show_password(self):
        print("Private Method - Password:", self.__password)

    # Public method to call private method
    def show_password(self):
        self.__show_password()



s1 = Student("Aniket", 85, "abc123")
s2 = Student("Rahul", 90, "xyz456")



student_ref = s1

print("Object Reference:")
print(student_ref)



print("\nPublic Variable:")
print(s1.name)


print("\nProtected Variable:")
print(s1._marks)




print("\nPrivate Variable:")



print("\nPublic Method:")
s1.display()



print("\nProtected Method:")
s1._show_marks()




print("\nPrivate Method:")
s1.show_password()



print("\n__str__() Method:")
print(s1)



s1.city = "Pune"

print("\nAttribute Created Outside Class:")
print(s1.city)



def display_student(student):
    print("\nFunction Receiving Object:")
    print("Name:", student.name)
    print("Marks:", student._marks)


display_student(s1)
display_student(s2)



print("\nMultiple Objects:")
print(s1)
print(s2)
```
