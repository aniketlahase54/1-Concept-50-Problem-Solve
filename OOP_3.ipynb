```python
# Create a User class with private __password. Create methods to change and verify the password.
class User:
    def __init__(self, password):
        self.__password = password

    def change_password(self, old_password, new_password):
        if self.__password == old_password:
            self.__password = new_password
            print("Password changed successfully")
        else:
            print("Incorrect old password")

    def verify_password(self, password):
        if self.__password == password:
            return True
        else:
            return False


u1 = User("abc123")

print(u1.verify_password("abc123"))

u1.change_password("abc123", "xyz789")

print(u1.verify_password("xyz789"))
print(u1.verify_password("abc123"))
```

```python
# Create a ComplexNumber class and overload + and -.
class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def __add__(self, other):
        return ComplexNumber(
            self.real + other.real,
            self.imag + other.imag
        )

    def __sub__(self, other):
        return ComplexNumber(
            self.real - other.real,
            self.imag - other.imag
        )

    def __str__(self):
        return f"{self.real} + {self.imag}i"


c1 = ComplexNumber(5, 3)
c2 = ComplexNumber(2, 1)

c3 = c1 + c2
c4 = c1 - c2

print(c3)
print(c4)
```

```python
# Final OOP Challenge: Build a mini Banking System using all the concepts below:
# Encapsulation
#     ↓
# Getter + Setter
#     ↓
# Inheritance
#     ↓
# Single / Multilevel / Multiple / Hierarchical
#     ↓
# super()
#     ↓
# Method Overriding
#     ↓
# Polymorphism
#     ↓
# Operator Overloading
#     ↓
# Duck Typing
#     ↓
# Abstraction


from abc import ABC, abstractmethod


class BankAccount(ABC):
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Invalid balance")

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: ₹{amount}")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrawn: ₹{amount}")
        else:
            print("Insufficient balance or invalid amount")

    @abstractmethod
    def account_type(self):
        pass

    def __add__(self, other):
        return self.__balance + other.get_balance()


class Customer(BankAccount):
    def __init__(self, name, balance):
        super().__init__(name, balance)

    def account_type(self):
        print("Customer Account")


class SavingsAccount(Customer):
    def __init__(self, name, balance, interest_rate):
        super().__init__(name, balance)
        self.interest_rate = interest_rate

    def account_type(self):
        print("Savings Account")

    def withdraw(self, amount):
        print("Savings Account Withdrawal")
        super().withdraw(amount)


class CurrentAccount(Customer):
    def account_type(self):
        print("Current Account")

    def withdraw(self, amount):
        print("Current Account Withdrawal")
        super().withdraw(amount)


class Loan:
    def loan_details(self):
        print("Loan facility available")


class PremiumAccount(SavingsAccount, Loan):
    def premium_service(self):
        print("Premium banking service")


class Employee:
    def employee_details(self):
        print("Bank Employee")


class Manager(Employee):
    def employee_details(self):
        print("Bank Manager")


class Clerk(Employee):
    def employee_details(self):
        print("Bank Clerk")


def show_account(account):
    account.account_type()


def process_transaction(account):
    account.withdraw(500)


def read_account(account):
    account.account_type()


s1 = SavingsAccount("Aniket", 50000, 6.5)
c1 = CurrentAccount("Rahul", 80000)

print("Name:", s1.name)
print("Balance:", s1.get_balance())

s1.set_balance(60000)
print("Updated Balance:", s1.get_balance())

s1.deposit(5000)
s1.withdraw(3000)

show_account(s1)
show_account(c1)

process_transaction(s1)
process_transaction(c1)

print("Combined Balance:", s1 + c1)

p1 = PremiumAccount("Amit", 100000, 7.0)

p1.account_type()
p1.loan_details()
p1.premium_service()

m1 = Manager()
cl1 = Clerk()

m1.employee_details()
cl1.employee_details()

read_account(s1)
read_account(c1)
```
