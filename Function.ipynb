```python
# 4️⃣ Variable-Length Arguments — *args — Q23–30
# Create a function using *args that prints all received numbers.
def num(*args):
    print(args)

num(1,2,3,4,5)    
```

```python
# Create sum_numbers(*args) that returns the sum.
def sum_num(*args):
    return sum(args)

print(sum_num(1, 2, 3, 4))
```

```python
# Create multiply_numbers(*args) that returns the product.
def multiply_numbers(*args):
    result = 1
    for num in args:
        result *= num
    return result

print(multiply_numbers(2, 3, 4))
```

```python
# Create find_max(*args) that returns the largest number.
def find_max(*args): 
    return max(args)


print(find_max(2,5,3,4,7,8))    
```

```python
# Create find_min(*args) that returns the smallest number.
def find_min(*args): 
    return min(args)


print(find_min(2,5,3,4,7,8)) 
```

```python
# Create average(*args) that returns the average.
def avg(*args): 
    return (sum(args)/len(args))


print(avg(2,3,4,5))    
```

```python
# Create count_items(*args) that returns the number of arguments.
def count_items(*args): 
    count = 0
    for num in args: 
        count+=1

    return count
print(count_items(1,2,3,4,5,6,7,8,9,0))    
```

```python
# Create print_names(*args) that prints every name received.
def print_names(*args): 
    for name in args: 
        print("name:",name)

print_names("Aniket","Om","kartik","Saurav")   
```

```python
# Use the nonlocal keyword to modify an enclosing variable.

# def outer():
#     message = "Hello"


#     def inner():
#         print(message)


#     inner()


# outer()

def outer():
    message = "Hello"

    def inner():
        nonlocal message
        message = "Hello Aniket"
        print(message)

    inner()
    print("Outer:", message)


outer()
```

```python
# 🚀 Mini Project — Student Result System

# Create functions using different argument types:

# 1. add_student()
# 2. calculate_total()
# 3. calculate_average()
# 4. calculate_grade()
# 5. display_student()




# 🚀 Mini Project — Student Result System

# 1. add_student() → **kwargs
def add_student(**kwargs):
    return kwargs


# 2. calculate_total() → *args
def calculate_total(*marks):
    return sum(marks)


# 3. calculate_average() → *args
def calculate_average(*marks):
    return sum(marks) / len(marks)


# 4. calculate_grade() → normal argument
def calculate_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


# 5. display_student() → **kwargs
def display_student(**student):
    print("\n--- Student Result ---")
    for key, value in student.items():
        print(key, ":", value)


# Student information
student = add_student(
    name="Aniket",
    roll_no=101,
    python=85,
    sql=90,
    power_bi=80
)

# Get marks
python = student["python"]
sql = student["sql"]
power_bi = student["power_bi"]

# Calculations
total = calculate_total(python, sql, power_bi)
average = calculate_average(python, sql, power_bi)
grade = calculate_grade(average)

# Display result
display_student(
    name=student["name"],
    roll_no=student["roll_no"],
    total=total,
    average=average,
    grade=grade
)
```
