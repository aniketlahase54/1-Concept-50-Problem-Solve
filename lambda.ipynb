```python
# 🚀 Mini Project: Create a student result system using lambda, map(), filter(), sorted(), and reduce().



students = [
    {"name": "Aniket", "marks": [85, 90, 78, 88, 92]},
    {"name": "Rahul", "marks": [65, 70, 68, 72, 60]},
    {"name": "Amit", "marks": [92, 95, 89, 94, 90]},
    {"name": "Sneha", "marks": [75, 80, 78, 82, 85]},
    {"name": "Priya", "marks": [45, 50, 42, 48, 55]}
]

# 1. Calculate total using reduce()
for student in students:
    student["total"] = reduce(lambda a, b: a + b, student["marks"])

# 2. Calculate average using lambda
for student in students:
    student["average"] = lambda total: total / 5

for student in students:
    student["average"] = student["average"](student["total"])

# 3. Calculate percentage using lambda
percentage = lambda total: total / 5

for student in students:
    student["percentage"] = percentage(student["total"])

# 4. Calculate grade using lambda
grade = lambda p: (
    "A+" if p >= 90 else
    "A" if p >= 80 else
    "B" if p >= 70 else
    "C" if p >= 60 else
    "F"
)

for student in students:
    student["grade"] = grade(student["percentage"])

# 5. Filter passed students
passed_students = list(
    filter(lambda student: student["percentage"] >= 40, students)
)

# 6. Sort students by percentage
ranked_students = sorted(
    passed_students,
    key=lambda student: student["percentage"],
    reverse=True
)

# 7. Display result
for rank, student in enumerate(ranked_students, start=1):
    print(
        rank,
        student["name"],
        student["total"],
        student["percentage"],
        student["grade"]
    )
```
